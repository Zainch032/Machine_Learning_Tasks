#### Lab Task 1:

```python
def grade(marks):
    if marks < 50:
        return "Grade F"
    elif 50 <= marks <= 60:
        return "Grade E"
    elif 61 <= marks <= 70:
        return "Grade D"
    elif 71 <= marks <= 80:
        return "Grade C"
    elif 81 <= marks <= 90:
        return "Grade B"
    elif 91 <= marks <= 100:
        return "Grade A"
    else:
        return "Invalid input! Marks should be between 1 and 100."

n = int(input("Enter your marks: "))
print(grade(n))
```

#### Lab task 2:

```python
def fact(n):
    if n == 0:
        return 1
    else:
        return n * fact(n-1)
    
fact(10)
```

#### Lab Task 3:


```python
def fib(n):
    if n <= 1:
        return n
    else:
        return fib(n-1) + fib(n-2)
    
fib(10)
```
