### Activity 13

```python
isPrime =True
i = 2
n = int(input("Enter a number"))
while i <= n:
    remainder = n % i
    if remainder == 0:
        isPrime = False
        break
    else:
        i += 1
        
if isPrime:
    print("it is a prime number")
else:
    print("it is not a prime number")
```
